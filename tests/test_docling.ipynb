element: In each iteration, this variable holds a specific Document Item. Depending on the PDF's content, this object will change its type:

TextItem: Standard sentences or paragraphs.

TableItem: A structural table object (which you can export to Markdown/DataFrames).

PictureItem: An image or figure found in the PDF.

SectionHeaderItem: A heading (e.g., "# Introduction").

_level: This integer represents the nesting depth of the element in the document's outline. For example, a main title might be at level 0, while a sub-bullet point inside a list might be at level 3.

.iterate_items(): This method is an iterator that yields every piece of the document in the exact order it appears visually. It ensures that your RAG system processes the "Abstract" before the "Conclusion," even if the PDF's internal metadata is scrambled.

```python
from docling_core.transforms.chunker.hybrid_chunker import HybridChunker
from docling_core.transforms.chunker.tokenizer.base import BaseTokenizer
from docling_core.transforms.chunker.tokenizer.huggingface import HuggingFaceTokenizer
from transformers import AutoTokenizer

EMBED_MODEL_ID = "sentence-transformers/all-MiniLM-L6-v2"

tokenizer: BaseTokenizer = HuggingFaceTokenizer(
    tokenizer=AutoTokenizer.from_pretrained(EMBED_MODEL_ID),
)
chunker = HybridChunker(tokenizer=tokenizer)
```

```python
chunks = chunker.chunk(document)
```

```python
final_langchain_docs = []
for chunk in chunks:
    # chunk.text already includes the hierarchical context
    doc = Document(
        page_content=chunk.text,
        metadata={
            "headings": chunk.meta.headings,
            "doc_items": [item.label for item in chunk.meta.doc_items]
        }
    )
    final_langchain_docs.append(doc)
```
